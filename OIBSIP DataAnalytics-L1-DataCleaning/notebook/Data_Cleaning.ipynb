## Missing Value Analysis

Missing values can affect calculations and reduce the quality of analysis. Therefore, missing values were identified and examined before applying any cleaning strategy.

The missing-value count was calculated for each column using `isnull().sum()`.

### Handling Missing Age Values

The `Age` column contains missing numerical values.

The median was selected to fill the missing Age values because it is less sensitive to extreme values than the mean. This helps preserve the overall distribution of the variable without allowing unusually high or low ages to strongly influence the replacement value.

```python
age_median = df['Age'].median()

df['Age'] = df['Age'].fillna(age_median)

print("Median Age used:", age_median)
print("Remaining missing Age values:", df['Age'].isnull().sum())
```

```python
df['Embarked'].value_counts()
```

### Handling Missing Embarked Values

The `Embarked` column is categorical and contains a small number of missing values.

The mode was selected because it represents the most frequently occurring category in the dataset. This is appropriate for filling a small number of missing categorical values while maintaining the existing category distribution.

```python
embarked_mode = df['Embarked'].mode()[0]

df['Embarked'] = df['Embarked'].fillna(embarked_mode)

print("Mode used:", embarked_mode)
print("Remaining missing Embarked values:", df['Embarked'].isnull().sum())
```

### Handling Missing Cabin Values

The `Cabin` column contains a substantial number of missing values.

Instead of replacing the missing cabin numbers with an artificial value, a new binary column named `Cabin_Unknown` was created. It indicates whether cabin information was missing.

The original `Cabin` column is retained because its available information may still be useful for analysis.

```python
df['Cabin_Unknown'] = df['Cabin'].isnull().astype(int)

df[['Cabin', 'Cabin_Unknown']].head()
```

## Outlier Detection Using IQR

Outliers are observations that are unusually far from the majority of values.

The Interquartile Range (IQR) method was used to identify potential outliers in numerical columns.

The IQR is calculated as:

IQR = Q3 - Q1

Lower Bound = Q1 - 1.5 × IQR

Upper Bound = Q3 + 1.5 × IQR

Potential outliers were identified for further investigation rather than automatically deleted.

```python
Q1_age = df['Age'].quantile(0.25)
Q3_age = df['Age'].quantile(0.75)

IQR_age = Q3_age - Q1_age

lower_age = Q1_age - 1.5 * IQR_age
upper_age = Q3_age + 1.5 * IQR_age

age_outliers = df[
    (df['Age'] < lower_age) |
    (df['Age'] > upper_age)
]

print("Age lower bound:", lower_age)
print("Age upper bound:", upper_age)
print("Number of Age outliers:", len(age_outliers))
```

```python
Q1_fare = df['Fare'].quantile(0.25)
Q3_fare = df['Fare'].quantile(0.75)

IQR_fare = Q3_fare - Q1_fare

lower_fare = Q1_fare - 1.5 * IQR_fare
upper_fare = Q3_fare + 1.5 * IQR_fare

fare_outliers = df[
    (df['Fare'] < lower_fare) |
    (df['Fare'] > upper_fare)
]

print("Fare lower bound:", lower_fare)
print("Fare upper bound:", upper_fare)
print("Number of Fare outliers:", len(fare_outliers))
```

### Outlier Handling Decision

Potential outliers were identified using the IQR method.

The identified observations were not automatically removed because statistical outliers may represent genuine observations rather than data-entry errors.

Therefore, the detected outliers were retained unless there was clear evidence that they were invalid values.
